```python
import numpy as np
import pandas as pd
import sklearn
```

```python
data=pd.DataFrame({'Price_USA' : [80,50,140,25],
                    'Price_INR' : [5000,7000,12000,3000]})
data
```

```python
#we will perform mathematiical wise
data['Price_USA_mimmax']=(data['Price_USA']-data['Price_USA'].min())/(data['Price_USA'].max()-data['Price_USA'].min())
data
```

```python
#we will perform mathematiical wise
data['Price_INR_mimmax']=(data['Price_INR']-data['Price_INR'].min())/(data['Price_INR'].max()-data['Price_INR'].min())
data
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
min_max = MinMaxScaler()
```

```python
#we need to use fit_transform ()
data['Price_USA_minsk'] = min_max.fit_transform(data[['Price_USA']])# extra brace for two dimension
data
```

```python
data['Price_USA_std'] = (data['Price_USA']-data['Price_USA'].mean())/(data['Price_USA'].std())
data
```

```python
#we need to use fit_transform ()
data['Price_INR_minsk'] = min_max.fit_transform(data[['Price_INR']])# extra brace for two dimension
data
```

```python
scaler = StandardScaler()
```

```python
data['Price_USA_scaler'] = scaler.fit_transform(data[['Price_USA']])
data
```

```python
data['Price_INR_scaler'] = scaler.fit_transform(data[['Price_INR']])
data
```

```python
from sklearn.preprocessing import RobustScaler #robust is used to standardize the values when the data is having many outlets
```

```python
robust=RobustScaler()
```

```python
data['Price_USA_robust']=robust.fit_transform(data[['Price_USA']])
data['Price_USA_robust']
data['Price_INR_robust']=robust.fit_transform(data[['Price_INR']])
data['Price_INR_robust']
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import OneHotEncoder, PolynomialFeatures
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
#loading the data
```

```python
data = fetch_ucirepo(id=9)
data
x=data.data.features.copy() # returns allnumerical features
y=data.data.targets['mpg']
```

```python
x.describe();
x.info()
```

```python
x['origin'].unique();
x['horsepower'].unique()
```

```python
#first we will transform horsepower coloumn and 
x['horsepower']=pd.to_numeric(x['horsepower'],errors='coerce')
x['horsepower'].info()
x['origin'] = x['origin'].astype(str)
```

```python
x.info();
#exit all numerical couloums into single
numerical_features=['cylinders','displacement','horsepower','weight','acceleration','model_year']
categorical_features=['origin']
x=x[numerical_features+categorical_features];
x.info();
x.isnull().sum()
x.shape
```

```python
#step3 quick eda (seaborn)
df=x.copy()
df['mpg']=y
#here we are mapping input and target values
df
```

```python
# Plot using seaborn's subplot() concept to divide the plot into multiple plots
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
# Plot graphs as per axes
sns.histplot(df['mpg'], kde=True, ax=axes[0], color='royalblue')
# Similar way we display other graphs in same plot
axes[0].set_title("MPG Distribution")
sns.scatterplot(x=df['weight'], y=df['mpg'], ax=axes[1], color='darkorange')
axes[1].set_title("Weight vs MPG Distribution")
sns.scatterplot(x=df['horsepower'], y=df['mpg'], ax=axes[2], color='mediumseagreen')
axes[2].set_title("HorsePower vs MPG Distribution")
plt.tight_layout()
plt.show()
```

```python
#correlation between our variables
plt.figure(figsize=(16,4))
sns.heatmap(df.select_dtypes(include=np.number).corr(),annot=True,
           cmap='coolwarm',
           center=0)
plt.title("correlation")
plt.tight_layout()
plt.show()
```

```python
print(x)
print(y)
#split rule -->70-30,75-25,80-20
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print("\n train test split")
print("training data",len(x_train))
print("testing data",len(x_test))
```

```python
#step5 : create a base line model
def evaluate(model_name,actual,predicted):
    """returns MAE,MSE,RMSE,R2-SCORE"""
    mse=mean_squared_error(actual,predicted)
    mae=mean_absolute_error(actual,predicted)
    return {'Model_Name':model_name,
    'MAE' :mean_absolute_error,
    'MSE':mse,
    'RMSE':np.sqrt(mse),
    'R2-score':r2_score(actual,predicted)}
results=[]
           
    
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
#Step-6 : Baseline model --> It prediicts the training set for every value
baseline_predictions = np.full(len(y_test),y_train.mean())
#baseline_predictions
#in above case we are keeping a baseline prediction'
results.append(evaluate("Mean baseline",y_test,baseline_predictions))
results
```

```python
#step 7 linear regression
simple_model=LinearRegression()
simple_model.fit(x_train[['horsepower']],y_train)
simple_model_predictions=simple_model.predict(x_test[['horsepower']])
results.append(evaluate("Simple Linear Regression:horsepower",y_test,simple_model_predictions))
print(len(results))
```
